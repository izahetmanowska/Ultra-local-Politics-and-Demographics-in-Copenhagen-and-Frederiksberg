```python
import pandas as pd

path = '../../../raw-data/election-2025/Kommunalvalg_2025_Frederiksberg_25-11-2025 14.54.22.csv'

votes_frb = pd.read_csv(path, sep=';')

print(votes_frb.columns)

votes_frb.head()
```

```python
unique_navn = votes_frb['Navn'].unique()
print(unique_navn)
```

```python
unique_listenavn = votes_frb['Listenavn'].unique()
print(unique_listenavn)
```

```python
path = '../../../raw-data/social-media/parties_info_frb.csv'

parties_info_frb = votes_frb[['Bogstavbetegnelse', 'Listenavn']].drop_duplicates()
parties_info_frb['Instagram Tag'] = ""
parties_info_frb['Follower Count'] = None
parties_info_frb['Number of Posts'] = None

parties_info_frb.to_csv(path, index=False,sep=';')
```

```python
path = '../../../raw-data/social-media/politicians_info_frb.csv'

politicians_info_frb = votes_frb[['Bogstavbetegnelse', 'Listenavn','Navn']].drop_duplicates()
politicians_info_frb = politicians_info_frb[politicians_info_frb['Navn'] != 'Listestemmer']
politicians_info_frb['Instagram Tag'] = ""
politicians_info_frb['Follower Count'] = None
politicians_info_frb['Number of Posts'] = None

politicians_info_frb.to_csv(path, index=False,sep=';')
```
