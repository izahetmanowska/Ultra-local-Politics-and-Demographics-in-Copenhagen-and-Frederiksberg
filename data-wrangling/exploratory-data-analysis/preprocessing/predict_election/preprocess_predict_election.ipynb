# Preprocessing data for predicting the election

```python
#Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
from pathlib import Path
from sklearn.preprocessing import LabelEncoder
import preprocess_predict_election as pppe


# Configuration
from preprocess_predict_election import AGE_GROUPS, EDUCATION_GROUPS, CAR_GROUP, BENEFIT_GROUP, KEEP_PARTIES
from pathlib import Path


CWD = Path.cwd()
PROJECT_ROOT = CWD.parents[2]

DEM_DATA_DIR = PROJECT_ROOT / "processed-data" / "demographics"
ELEC_DATA_DIR = PROJECT_ROOT / "processed-data" / "elections"

OUTPUT_DIR = PROJECT_ROOT / "processed-data" / "combined"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Getting the demografic data and conforming it into pandas dataframes.

# 2009
path = DEM_DATA_DIR / "relative_2009.csv"

df_dem_2009 = pd.read_csv(path, sep=',')

# 2013
path = DEM_DATA_DIR / "relative_2013.csv"

df_dem_2013 = pd.read_csv(path, sep=',')

# 2017 
path = DEM_DATA_DIR / "relative_2017.csv"

df_dem_2017 = pd.read_csv(path, sep=',')

# 2021 
path = DEM_DATA_DIR / "relative_2021.csv"

df_dem_2021 = pd.read_csv(path, sep=',')

```

```python
#Normalizing income (globally/for all years) using log transform and min-max scaling to get 0-100 range
# The dict for the dfs
dfs_dict = {
    2009: df_dem_2009,
    2013: df_dem_2013,
    2017: df_dem_2017,
    2021: df_dem_2021
}


dfs_normalized = pppe.normalize_income(dfs_dict)

# Extracting back to individual variables in year df's
df_dem_2009 = dfs_normalized[2009]
df_dem_2013 = dfs_normalized[2013]
df_dem_2017 = dfs_normalized[2017]
df_dem_2021 = dfs_normalized[2021]
```

```python
#Preprocessing the demografic data for the question

df_dem_2009_pp = pppe.preprocess_demographics(df_dem_2009)

df_dem_2013_pp = pppe.preprocess_demographics(df_dem_2013)

df_dem_2017_pp = pppe.preprocess_demographics(df_dem_2017)

df_dem_2021_pp = pppe.preprocess_demographics(df_dem_2021)

```

```python
# 2009
path = ELEC_DATA_DIR / "relative_2009.csv"

df_elec_2009 = pd.read_csv(path, sep=',')

# 2013
path = ELEC_DATA_DIR / "relative_2013.csv"

df_elec_2013 = pd.read_csv(path, sep=',')

# 2017 
path = ELEC_DATA_DIR / "relative_2017.csv"

df_elec_2017 = pd.read_csv(path, sep=',')

# 2021 
path = ELEC_DATA_DIR / "relative_2021.csv"

df_elec_2021 = pd.read_csv(path, sep=',')

# 2021 
path = ELEC_DATA_DIR / "relative_2025.csv"

df_elec_2025 = pd.read_csv(path, sep=',')
```

```python
party_cols_2009 = pppe.get_party_columns(df_elec_2009, start_after='Valid votes')
party_cols_2013 = pppe.get_party_columns(df_elec_2013, start_after='Valid votes')
party_cols_2017 = pppe.get_party_columns(df_elec_2017, start_after='Valid votes')
party_cols_2021 = pppe.get_party_columns(df_elec_2021, start_after='Valid votes')
party_cols_2025 = pppe.get_party_columns(df_elec_2025, start_after='Valid votes')

print(f"Parties in 2009 {party_cols_2009}")
print(f"Parties in 2013 {party_cols_2013}")
print(f"Parties in 2017 {party_cols_2017}")
print(f"Parties in 2021 {party_cols_2021}")
print(f"Parties in 2025 {party_cols_2025}")
```

```python
df_elec_2009_pp = pppe.preprocess_election(df_elec_2009)

df_elec_2013_pp = pppe.preprocess_election(df_elec_2013)

df_elec_2017_pp = pppe.preprocess_election(df_elec_2017)

df_elec_2021_pp = pppe.preprocess_election(df_elec_2021)

df_elec_2025_pp = pppe.preprocess_election(df_elec_2025)
```

```python
winning_parties_2009 = df_elec_2009_pp["winning_party"].unique()
winning_parties_2013 = df_elec_2013_pp["winning_party"].unique()
winning_parties_2017 = df_elec_2017_pp["winning_party"].unique()
winning_parties_2021 = df_elec_2021_pp["winning_party"].unique()
winning_parties_2025 = df_elec_2025_pp["winning_party"].unique()


print(f"Winning parties in 2009 {winning_parties_2009}")
print(f"Winning parties in 2013 {winning_parties_2013}")
print(f"Winning parties in 2017 {winning_parties_2017}")
print(f"Winning parties in 2021 {winning_parties_2021}")
print(f"Winning parties in 2025 {winning_parties_2025}")
```
