# Feature Selection for Predicting the Election

This section concerns feature selection nescessary for answering the question: 

- Which demographic features best predict vote shares for major parties in Copenhagen and Frederiksberg, and can these patterns forecast the 2025 election winner?

It contains:

- Feature selection through correlation analysis
    - dropping features that are highly correlated
    
- Selecting top 10 features through ElasticNet model


```python
#Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
import feature_select_predict_election as fs
from pathlib import Path

#Configuration
from feature_select_predict_election import EXCLUDE_COLS

CWD = Path.cwd()
PROJECT_ROOT = CWD.parents[3]

DATA_DIR = PROJECT_ROOT / "processed-data" / "combined"

OUTPUT_DIR = PROJECT_ROOT / "data-wrangling"/ "preprocessing"/ "predict_election"/"feature_selection"/ "pe_heatmaps"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Getting the combined data and conforming it into pandas dataframes.

# 2009
path = DATA_DIR / "pp_combined_2009.csv"

df_2009 = pd.read_csv(path, sep=',')

# 2013
path = DATA_DIR / "pp_combined_2013.csv"

df_2013 = pd.read_csv(path, sep=',')

# 2017 
path = DATA_DIR / "pp_combined_2017.csv"

df_2017 = pd.read_csv(path, sep=',')

# 2021 
path = DATA_DIR / "pp_combined_2021.csv"

df_2021 = pd.read_csv(path, sep=',')
```

### Conclusion from correlation analysis

From the correlation analysis it is evident that some feature pairs are very highly correlated across election years. Here especially:

- Seniors_65_plus and Benefit type_Folkepension (0.997-0.999)
- Socioeconomy_Top executives and Income_Gross income top20pct (0.91-0.93)
- Housing tenure_People_Ejer and Households_with_car(0.81-0.89)

Common for the pairs are that they all capture the same underlying things. Therefor I will drop the features: 'Benefit type_Folkepension', 'Income_Gross income top20pct', 'Households_with_car'.

The other correlated features, they are still very correlated, however they correlation does not capture the same thing or if it does it is because it is negative and it captures that when you have fewer of one group you often have more of another.
These pairs are the following:      
- Education_Primary_lower_secondary & Education_Long_higher_education   
- People_on_benefits & Education_Primary_lower_secondary   
- People_on_benefits & Housing tenure_People_Almen   
- Education_Vocational_training & Education_Long_higher_education   
- Education_Primary_lower_secondary & Housing tenure_People_Almen   
- People_on_benefits & Socioeconomy_Top executives
