Notebook for creating relative_2009, where all data is % of the different totals for easier comparison across polling areas (instead of absolute numbers).


# Initial Steps

```python
import pandas as pd 

path = "../../../processed-data/demographics/absolute_2009.csv"

df = pd.read_csv(path)
df.head()
```

```python
print(df.columns)
```

# Calculations

The results of the following will be floats, and omit "%" sign at the end of each value.

'Citizenship_ 'Housing type_People_ 'Benefit type_ 'Sex_ 'Age_' with 'Population' as the total to get the relative


```python
df['Sex_Female'] = (df['Sex_Female']/df['Population'])*100
df['Sex_Male'] = (df['Sex_Male']/df['Population'])*100

df['Age_0-4 years'] = (df['Age_0-4 years']/df['Population'])*100
df['Age_10-14 years'] = (df['Age_10-14 years']/df['Population'])*100
df['Age_15-17 years'] = (df['Age_15-17 years']/df['Population'])*100
df['Age_18-19 years'] = (df['Age_18-19 years']/df['Population'])*100
df['Age_20-24 years'] = (df['Age_20-24 years']/df['Population'])*100
df['Age_25-29 years'] = (df['Age_25-29 years']/df['Population'])*100
df['Age_30-34 years'] = (df['Age_30-34 years']/df['Population'])*100
df['Age_35-39 years'] = (df['Age_35-39 years']/df['Population'])*100
df['Age_40-44 years'] = (df['Age_40-44 years']/df['Population'])*100
df['Age_45-49 years'] = (df['Age_45-49 years']/df['Population'])*100
df['Age_5-9 years'] = (df['Age_5-9 years']/df['Population'])*100
df['Age_50-54 years'] = (df['Age_50-54 years']/df['Population'])*100
df['Age_55-59 years'] = (df['Age_55-59 years']/df['Population'])*100
df['Age_60-64 years'] = (df['Age_60-64 years']/df['Population'])*100
df['Age_65-69 years'] = (df['Age_65-69 years']/df['Population'])*100
df['Age_70- years'] = (df['Age_70- years']/df['Population'])*100

df['Citizenship_Africa'] = (df['Citizenship_Africa']/df['Population'])*100
df['Citizenship_Asia and Oceania'] = (df['Citizenship_Asia and Oceania']/df['Population'])*100
df['Citizenship_Denmark'] = (df['Citizenship_Denmark']/df['Population'])*100
df['Citizenship_Former Yugoslavia'] = (df['Citizenship_Former Yugoslavia']/df['Population'])*100
df['Citizenship_New EU countries'] = (df['Citizenship_New EU countries']/df['Population'])*100
df['Citizenship_Nordic'] = (df['Citizenship_Nordic']/df['Population'])*100
df['Citizenship_North America'] = (df['Citizenship_North America']/df['Population'])*100
df['Citizenship_Old EU countries'] = (df['Citizenship_Old EU countries']/df['Population'])*100
df['Citizenship_Other Europe'] = (df['Citizenship_Other Europe']/df['Population'])*100
df['Citizenship_South and Central America'] = (df['Citizenship_South and Central America']/df['Population'])*100
df['Citizenship_Turkey'] = (df['Citizenship_Turkey']/df['Population'])*100

df['Housing type_People_Apartment buildings'] = (df['Housing type_People_Apartment buildings']/df['Population'])*100
df['Housing type_People_Detached house'] = (df['Housing type_People_Detached house']/df['Population'])*100
df['Housing type_People_Terraced house'] = (df['Housing type_People_Terraced house']/df['Population'])*100

df['Benefit type_Foertidspension'] = (df['Benefit type_Foertidspension']/df['Population'])*100
df['Benefit type_Folkepension'] = (df['Benefit type_Folkepension']/df['Population'])*100
df['Benefit type_Kontanthjaelp'] = (df['Benefit type_Kontanthjaelp']/df['Population'])*100
df['Benefit type_Modtager ikke ydelser'] = (df['Benefit type_Modtager ikke ydelser']/df['Population'])*100
```

Socioeconomy_ with Socioeconomy_Total

```python
df['Socioeconomy_Self-employed'] = (df['Socioeconomy_Self-employed']/df['Socioeconomy_Total'])*100
df['Socioeconomy_Top executives'] = (df['Socioeconomy_Top executives']/df['Socioeconomy_Total'])*100
df['Socioeconomy_Unemployed'] = (df['Socioeconomy_Unemployed']/df['Socioeconomy_Total'])*100
```

Immigrants_ with Immigrants_Total

```python
df['Immigrants_Africa'] = (df['Immigrants_Africa']/df['Immigrants_Total'])*100
df['Immigrants_Asia and Oceania'] = (df['Immigrants_Asia and Oceania']/df['Immigrants_Total'])*100
df['Immigrants_Former Yugoslavia'] = (df['Immigrants_Former Yugoslavia']/df['Immigrants_Total'])*100
df['Immigrants_New EU countries'] = (df['Immigrants_New EU countries']/df['Immigrants_Total'])*100
df['Immigrants_Nordic'] = (df['Immigrants_Nordic']/df['Immigrants_Total'])*100
df['Immigrants_North America'] = (df['Immigrants_North America']/df['Immigrants_Total'])*100
df['Immigrants_Not specified'] = (df['Immigrants_Not specified']/df['Immigrants_Total'])*100
df['Immigrants_Old EU countries'] = (df['Immigrants_Old EU countries']/df['Immigrants_Total'])*100
df['Immigrants_Other Europe'] = (df['Immigrants_Other Europe']/df['Immigrants_Total'])*100
df['Immigrants_South and Central America'] = (df['Immigrants_South and Central America']/df['Immigrants_Total'])*100
df['Immigrants_Turkey'] = (df['Immigrants_Turkey']/df['Immigrants_Total'])*100
```

Education_ with Education_Total

```python
df['Education_18-29 years_Long higher education'] = (df['Education_18-29 years_Long higher education']/df['Education_Total'])*100
df['Education_18-29 years_Not specified'] = (df['Education_18-29 years_Not specified']/df['Education_Total'])*100
df['Education_18-29 years_Primary and lower secondary'] = (df['Education_18-29 years_Primary and lower secondary']/df['Education_Total'])*100
df['Education_18-29 years_Vocational training'] = (df['Education_18-29 years_Vocational training']/df['Education_Total'])*100
df['Education_30-70+ years_Long higher education'] = (df['Education_30-70+ years_Long higher education']/df['Education_Total'])*100
df['Education_30-70+ years_Not specified'] = (df['Education_30-70+ years_Not specified']/df['Education_Total'])*100
df['Education_30-70+ years_Primary and lower secondary'] = (df['Education_30-70+ years_Primary and lower secondary']/df['Education_Total'])*100
df['Education_30-70+ years_Vocational training'] = (df['Education_30-70+ years_Vocational training']/df['Education_Total'])*100
df['Education_Total 18-29 years'] = (df['Education_Total 18-29 years']/df['Education_Total'])*100
df['Education_Total 30-70+ years'] = (df['Education_Total 30-70+ years']/df['Education_Total'])*100
```

Housing tenure_People_ with Housing tenure_Total people

```python
df['Housing tenure_People_Almen'] = (df['Housing tenure_People_Almen']/df['Housing tenure_Total people'])*100
df['Housing tenure_People_Andel'] = (df['Housing tenure_People_Andel']/df['Housing tenure_Total people'])*100
df['Housing tenure_People_Ejer'] = (df['Housing tenure_People_Ejer']/df['Housing tenure_Total people'])*100
df['Housing tenure_People_Private rental'] = (df['Housing tenure_People_Private rental']/df['Housing tenure_Total people'])*100
```

Housing tenure_Homes_ with Housing tenure_Total homes

```python
df['Housing tenure_Homes_Almen'] = (df['Housing tenure_Homes_Almen']/df['Housing tenure_Total homes'])*100
df['Housing tenure_Homes_Andel'] = (df['Housing tenure_Homes_Almen']/df['Housing tenure_Total homes'])*100
df['Housing tenure_Homes_Ejer'] = (df['Housing tenure_Homes_Almen']/df['Housing tenure_Total homes'])*100
df['Housing tenure_Homes_Private rental'] = (df['Housing tenure_Homes_Almen']/df['Housing tenure_Total homes'])*100
```

Housing type_Homes with Housing type_Total homes

```python
df['Housing type_Homes_Apartment buildings'] = (df['Housing type_Homes_Apartment buildings']/df['Housing type_Total homes'])*100
df['Housing type_Homes_Detached house'] = (df['Housing type_Homes_Detached house']/df['Housing type_Total homes'])*100
df['Housing type_Homes_Terraced house'] = (df['Housing type_Homes_Terraced house']/df['Housing type_Total homes'])*100
```

Car_ with (pre-calculated) total households from all three car columns

```python
df['Car total_households'] = (df['Car_Households with 1 car']+df['Car_Households with 2+ cars']+df['Car_Households with no car'])
```

```python
df['Car_Households with 1 car'] = (df['Car_Households with 1 car']/df['Car total_households'])*100
df['Car_Households with 2+ cars'] = (df['Car_Households with 2+ cars']/df['Car total_households'])*100
df['Car_Households with no car'] = (df['Car_Households with no car']/df['Car total_households'])*100
```

Income_Median household income LEAVE AS IS

Income_Gross income top20pct with Income_Total households

```python
df['Income_Gross income top20pct'] = (df['Income_Gross income top20pct']/df['Income_Total households'])*100
```
