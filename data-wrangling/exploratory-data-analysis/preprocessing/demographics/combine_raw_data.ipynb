## PersonsBySocioeconomyIndustry

We want data for arbejdsløse, topledere and selvstændige. But there are similar column names for both topledere and selvstændige - we need to check these columns to see if data is duplicated, different, or only exists in one of the similarly named columns. 

```python
# Socio-økonomisk status og brancher fordelt på afstemningsområder,Population_PersonsBySocioeconomyIndustry.csv
path = PATH + 'PersonsBySocioeconomyIndustry.csv'
prefixes = [f"{PREFIX}{year}" for year in YEARS]
pattern = ' - Socio-økonomisk status og brancher fordelt på afstemningsområder_'
wanted_cols = [
    '07. Arbejdsløse_',
    '02. Topledere_',
    '01. Selvstændig og medhj._'
]
cols = []

df = pd.read_csv(path, sep=';', dtype=str)
df = df[df['Gruppe'].isin(CPH_AREAS)]

for prefix in prefixes:
    for name in wanted_cols:
        colname_start = f'{prefix}{pattern}{name}'
        for col in df.columns:
            if col.startswith(colname_start):
                cols.append(col)

final_cols = COMMON_COLS + cols
df = df[final_cols]

df_transposed = df.transpose()

df_transposed.tail()
```

## HousingType

DST says:

1. Enfamilie-, stue-, parcelhus (BBR-koder (110 Stuehus til landbrugsejendom og 120 Fritliggende enfamilieshus (parcelhus)).
2. Række-, kæde- dobbelthus (BBR-kode 130 Række-, kæde- eller dobbelthus (lodret adskillelse mellem enhederne)).
3. Etagebolig (BBR-kode 140 Etageboligbebyggelse (flerfamiliehus, herunder tofamiliehus (vandret adskillelse mellem enhederne))).
4. Kollegiebolig (BBR-kode 150 Kollegium)
6. Fritidshus (BBR-koder 510 Sommerhus. 540 Kolonihavehus.590 Anden enhed til fritidsformål)

In our data, we have:
- 01. Stuehuse og parcelhuse (detached houses/villas)
- 02. Række- og kædehuse (terraced houses)
- 03. Flerfamiliehuse (multi-family/block of flats)
- 04. Kollegier (student halls)
- 05. Sommerhuse (holiday homes)

So 'flerfamiliehuse' is apartment buildings/multi-storey residential buildings - or just two-family-houses, where the two homes are separated horizontally (as opposed to vertically, which would go under 'række- og kædehuse')
We will assume that 05. sommerhuse also covers kolonihavehuse, which we have a few of in cph.

I think we should put detached and terraced houses together in one column, blocks in one column. And ignore the rest.

There are duplicate 'Gruppe' values - we only want 58 rows, but we get 63, because we're merging on several common columns, and some of them are different. It is only the CSV 'PersonsBySocioeconomyIndustry.csv' that is causing some of them being different, I think. We'll merge just on 'Gruppe' instead, and then deal with duplicate common columns after.

```python
names = [
    'socio',
    'citizenship',
    'immigrants',
    'education',
    'tenure',
    'type',
    'simple',
    'builtyear'
]
years = [2009, 2013, 2017, 2021]

combined_dfs = {}
for year in years:
    # start with first dataframe
    df_combined = globals()[f'{names[0]}_{year}'].copy()
    # merge subsequent dataframes
    for name in names[1:]:
        df_name = f'{name}_{year}'
        df_combined = df_combined.merge(
            globals()[df_name], 
            on='Gruppe',
            how='outer',
            suffixes=('', f'_{name}')
        )
    combined_dfs[year] = df_combined

df_2009 = combined_dfs[2009]
df_2013 = combined_dfs[2013]
df_2017 = combined_dfs[2017]
df_2021 = combined_dfs[2021]

print(f"df_2021 shape: {df_2021.shape}")
print(f"Unique Gruppe values: {df_2021['Gruppe'].nunique()}")
```

### Add and drop geography columns
I want to remove StorKredsNr and LandsdelsNr, but add 'Valgsted navn', 'Kreds navn' and 'Kommune navn'. Preferably, the first six columns will be 'Gruppe', 'ValgstedId', 'Valgsted navn' 'KredsNr', 'Kreds navn' and 'Kommune navn', in that order.

```python
def add_geography_columns(df, geography_csv_path):
    """
    Add geography columns to dataframe by matching ValgstedId.
    
    Parameters:
    - df: DataFrame to add columns to
    - geography_csv_path: Path to Geography.csv
    
    Returns:
    - DataFrame with added geography columns
    """    
    # Read Geography.csv
    geo_df = pd.read_csv(geography_csv_path, sep=';', encoding='utf-8', quotechar='"')
    
    # Clean column names (remove any extra whitespace/quotes)
    geo_df.columns = geo_df.columns.str.strip().str.strip('"')

    # Filter for KommuneNr 101 or 147
    geo_df = geo_df[geo_df['KommuneNr'].astype(str).isin(['101', '147'])].copy()

    # Debug: print first few rows to see what we're working with
    print("Geography CSV columns:", geo_df.columns.tolist())
    print("First few Valgsted Ids:", geo_df['Valgsted Id'].head().tolist())

    # Create a dictionary for quick lookup
    geo_dict = {}
    for _, row in geo_df.iterrows():
        # Clean the ValgstedId (remove quotes, whitespace)
        valgsted_id = str(row['Valgsted Id']).strip().strip('"')
        geo_dict[valgsted_id] = {
            'Valgsted navn': str(row['Valgsted navn']).strip().strip('"'),
            'Kreds navn': str(row['Kreds navn']).strip().strip('"'),
            'Kommune navn': str(row['Kommune navn']).strip().strip('"')
        }
    
    print(f"Total entries in geo_dict: {len(geo_dict)}")
    print(f"Sample keys: {list(geo_dict.keys())[:5]}")
    
    # Create new columns
    df_result = df.copy()
    valgsted_navne = []
    kreds_navne = []
    kommune_navne = []
    
    for _, row in df_result.iterrows():
        valgsted_ids = str(row['ValgstedId']).split(';')
        
        # Collect names for each ValgstedId
        v_names = []
        k_names = []
        kom_names = []
        
        for vid in valgsted_ids:
            vid = vid.strip()
            if vid not in geo_dict:
                print(f"Missing ValgstedId: '{vid}' (length: {len(vid)})")
                print(f"Available keys starting with '101': {[k for k in geo_dict.keys() if k.startswith('101')][:10]}")
                raise ValueError(f"ValgstedId '{vid}' not found in Geography.csv")
            
            v_names.append(geo_dict[vid]['Valgsted navn'])
            k_names.append(geo_dict[vid]['Kreds navn'])
            kom_names.append(geo_dict[vid]['Kommune navn'])
        
        # Join with semicolons, remove duplicates while preserving order
        valgsted_navne.append(';'.join(v_names))
        
        # For Kreds and Kommune, use single value if all same, otherwise semicolon-separate
        if len(set(k_names)) == 1:
            kreds_navne.append(k_names[0])
        else:
            kreds_navne.append(';'.join(k_names))
            
        if len(set(kom_names)) == 1:
            kommune_navne.append(kom_names[0])
        else:
            kommune_navne.append(';'.join(kom_names))
    
    # Add the new columns
    df_result['Valgsted navn'] = valgsted_navne
    df_result['Kreds navn'] = kreds_navne
    df_result['Kommune navn'] = kommune_navne
    
    # Drop the columns we don't want
    df_result = df_result.drop(columns=['StorKredsNr', 'LandsdelsNr'])
    
    # Reorder columns to put the first 6 in the desired order
    first_cols = ['Gruppe', 'ValgstedId', 'Valgsted navn', 'KredsNr', 'Kreds navn', 'Kommune navn']
    other_cols = [col for col in df_result.columns if col not in first_cols]
    df_result = df_result[first_cols + other_cols]
    
    return df_result

geography_path = ROOT + 'raw-data/historical-elections/election-results/Geography.csv'

for year in years:
    combined_dfs[year] = add_geography_columns(combined_dfs[year], geography_path)
    print(f"Added geography columns to {year}")

# Reassign
df_2009, df_2013, df_2017, df_2021 = [combined_dfs[year] for year in years]

# Check the result
print("\nFirst few rows of df_2021:")
print(df_2021[['Gruppe', 'ValgstedId', 'Valgsted navn', 'KredsNr', 'Kreds navn', 'Kommune navn']].head(10))
```

## Convert (some) floats to ints

```python
def convert_floats_to_int(df):
    """Convert float columns to int if they have no fractional parts"""
    df_converted = df.copy()
    
    for col in df_converted.columns:
        # Check if column is float type
        if df_converted[col].dtype == 'float64':
            # Check if all non-null values have no fractional part
            non_null_values = df_converted[col].dropna()
            if len(non_null_values) > 0 and (non_null_values % 1 == 0).all():
                # Convert to Int64 (nullable integer type to preserve NaN if present)
                df_converted[col] = df_converted[col].astype('int64')
    
    return df_converted

for year in years:
    combined_dfs[year] = convert_floats_to_int(combined_dfs[year])
    print(f"Converted {year}")

print("\ndf_2021 dtypes before conversion:")
print(df_2021.dtypes.value_counts())

# Reassign
df_2009, df_2013, df_2017, df_2021 = [combined_dfs[year] for year in years]

# Check the result
print("\ndf_2021 dtypes after conversion:")
print(df_2021.dtypes.value_counts())
```

## Reorder and save to csv

```python
column_order_prefixes = [
    'Sex_',
    'Age_',
    'Citizenship_',
    'Income_',
    'Car_',
    'Crime_',
    'Benefit type_',
    'Housing type_',
    'Year built_',
    'Housing tenure_',
    'Education_',
    'Immigrants_',
    'Socioeconomy_'
]

for year in years:
    df = combined_dfs[year]
    
    # Start with the first 7 columns
    first_cols = ['Gruppe', 'ValgstedId', 'Valgsted navn', 'KredsNr', 'Kreds navn', 'Kommune navn', 'Population']
    ordered_cols = first_cols.copy()
    
    # Add columns in the specified prefix order
    for prefix in column_order_prefixes:
        matching_cols = [col for col in df.columns if col.startswith(prefix) and col not in ordered_cols]
        ordered_cols.extend(sorted(matching_cols))  # Sort alphabetically within each group
    
    # Add any remaining columns that weren't caught (shouldn't be any, but just in case)
    remaining_cols = [col for col in df.columns if col not in ordered_cols]
    if remaining_cols:
        print(f"Warning: {year} has unexpected columns: {remaining_cols}")
        ordered_cols.extend(remaining_cols)
    
    # Reorder the dataframe
    df = df[ordered_cols]
    combined_dfs[year] = df
    print(f"Reordered {year}: {df.shape}")

# Reassign
df_2009, df_2013, df_2017, df_2021 = [combined_dfs[year] for year in years]

# Save to CSVs
for year in years:
    output_path = f'{OUTPUT_PATH}absolute_{year}.csv'
    combined_dfs[year].to_csv(output_path, index=False)
    print(f"Saved {output_path}")

# Verify the column order
print("\nColumn order for df_2021:")
for i, col in enumerate(df_2021.columns[:20], 1):
    print(f"{i}. {col}")
```
