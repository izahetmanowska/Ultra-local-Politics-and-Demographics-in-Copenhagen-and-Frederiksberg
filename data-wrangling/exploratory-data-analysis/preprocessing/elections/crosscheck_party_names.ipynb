````python
PATH = ROOT + 'raw-data/historical-elections/election-results/'
filename = PATH + 'ElectionData.csv'
all_columns = pd.read_csv(filename, sep=';', nrows=0).columns.tolist()
cols = ['Gruppe', 'ValgstedId'] + [col for col in all_columns if col.startswith('KV2021')]
df = pd.read_csv(filename, sep=';', usecols=cols)
print(df.columns)
````
