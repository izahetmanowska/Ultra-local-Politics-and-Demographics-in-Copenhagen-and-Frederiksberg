# 05 Clustering

#### \#\#\# *Characterising neighbourhoods* \#\#\#

This notebook is for clustering methods, cluster characterisation and visualisations.

We first determine the optimal number of clusters $k$ using the elbow method which plots the within-cluster sum of squares (WCSS or inertia) across different values of $k$, but visually, the elbow in our plot is hard to spot. Instead, we used KneeLocator from the kneed library to determine the point, $k=5$.

Three clustering algorithms are implemented and compared: k-means clustering, hierarchical clustering (using Ward's linkage method), and Gaussian Mixture Models (GMM). All three are fitted to the five principal components, and for k-means and GMM, with $k=5$ clusters. The algorithms are evaluated using Silhouette Score (how similar is each point to its own cluster compared to other clusters?), the Davies-Bouldin Index (the average similarity ratio of each cluster with its most similar cluster, lower is better), and the Calinski-Harabasz Index (the ratio of between-cluster variance to within-cluster variance).

k-Means outperforms the other algorithms on all three metrics, but it is worth noting that a silhouette score of $0.324$ is not very high; scores close to 1 show well-defined clusters, 0 indicates cluster overlap, and scores close -1 shows misclassification. This tells us that even though we can divide polling areas in Copenhagen and Frederiksberg into clusters, the differences between the clusters maybe aren't that big, and there may be some overlap.

To characterise and interpret the five clusters identified by k-means, we calculate eta-squared values ($\eta^2$) for the original 45 variables to identify which demographic/socioeconomic features most strongly differentiate the clusters. Eta-squared provides a standardised measure of effect size, indicating the proportion of total variance in each variable that is explained by cluster membership: $\eta^2 = \frac{SS_{between}}{SS_{total}}$, where $SS_{between}$ is the sum of squares between clusters and $SS_{total}$ is the total sum of squares. The values range from 0 to 1; higher values indicate that cluster membership explains more of the variance in that variable, making the variable more useful for characterising and distinguishing clusters.

Some of the top features are then used for visualisations (heatmap and spider plots). The features reveal that education levels, immigration and citizenship (Western vs. Global South vs. Eastern Europe), housing characteristics (type, tenure, construction period), income and welfare benefits, and voter turnout best describe the differences between each cluster. We also identify descriptive characteristics for each cluster by examining how cluster means deviate from global means across all features. Clusters are then described based on their most distinctive demographic, socioeconomic, and housing characteristics.

We use two different methods to identify outlying polling areas. 

We apply DBSCAN (Density-Based Spatial Clustering of Applications with Noise) independently of the clustering with parameters $\epsilon = 3.6$ and \verb|minimum points = 3| to identify polling areas in low-density regions of the feature space. Among the found outliers are: 'Syd, Sundbyvester' (Ørestad), 'Sydhavn, Vesterbro' (new Sydhavn), and 'Nord, Brønshøj' (Tingbjerg). Interestingly, k-means with $k=5$ assigns the first two of these to their own small cluster, but they are prevented from that with DBSCAN when ´minimum points = 3`. This suggests that these two polling areas can form a small cluster together rather than being true outliers.

We also use k-means to identify outliers based on distance from cluster centroids. The polling areas furthest from their assigned cluster centroids were: 'Indre By, Indre By' (assigned to Cluster 4), 'Nord, Brønshøj' (assigned to Cluster 1) and 'Søndermark, Slots' (assigned to Cluster 2). These polling areas scored higher or lower on different demographic features than their own cluster's mean.

#### Findings

**Research questions:**

1. How do neighbourhoods in Copenhagen and Frederiksberg cluster based on demographic and socioeconomic characteristics?
2. What distinguishes these clusters?

With k-means, we identified five distinct neighbourhood 'types' in Copenhagen and Frederiksberg. 

**Cluster 0** is urban with well-educated young adults mixed with some poverty. It has relatively high shares of young adults, co-op housing, and young adults with long university degrees, but it also has higher shares of people on benefits such as 'kontanthjælp' and disability pension, which may indicate that the neighbourhoods in this cluster have not yet been completely gentrified. The polling areas in the cluster have fewer detached houses, fewer people that own the homes they live in, fewer top executives and newly built housing, lower income, and a low share of households with cars. It contains all of Nørrebro (except for central 'Øst, Nørrebro'), Amagerbro, some of Nordvest, a few polling areas in Valby and outer Østerbro, and a single area in Vesterbro.

**Cluster 1** is socially vulnerable with low income and education levels and high shares of social housing, people on welfare benefits and disability pension, and citizens from the Global South and Eastern Europe. It has low shares of Western immigrants and citizens, co-op housing, higher education, top executives and self-employed people. The cluster contains, among others, Tingbjerg, some other parts of Brønshøj Utterslev, old Sydhavn, parts of Amager (Urbanplanen), some outer parts of Valby (Folehaven and Vigerslev). These areas form a banana-shaped arc from northwestern to southern Copenhagen, which is reminiscent of the controversal term "Den Rådne Banan" (The Rotten Banana) once used to describe economically marginalised parts of peripheral Denmark.

'Nord, Brønshøj' - or Tingbjerg - is an outlier in this cluster. This is the only polling area which consists only of social housing. Here, the education levels are lower, more people are on benefits, the majority of people living here are immigrants from the Global South, the voter turnout is around 33 percent and only 50 percent have Danish citizenship. It's also a family neighbourhood with more children than other areas in the same cluster.

**Cluster 2** is more suburban with high shares of homeowners with multiple cars. It has high shares of detached houses, home ownership, people educated within skilled trades and households with two or more cars. It scores low on shares of apartments, co-op housing, new housing, Western citizens and immigrants, citizens from 'Other Europe' and people between 18-29 years whose highest education level is a long university degree. The cluster includes most of the villa-heavy polling areas in Brønshøj, Vanløse, outer parts of Frederiksberg, and some parts of Amager.

The polling area 'Søndermark, Slots' (around Roskildevej on outer Frederiksberg) is an outlier in this cluster. Relatively to other areas in the same cluster, it's less suburban with more apartments and less villas and row houses. Its has more elderly people on old-age pension and more females (likely old females who's outlived their husbands).

**Cluster 3** is primarily characterised by new, hyper-modern apartment housing, very high income and lots of internationals. It has high shares of people living in housing built after 2010, home ownership, Western citizens and citizens from Other Europe, young adults with long university degrees, and high income. It scores low old housing and co-op housing, detached houses, people on benefits, and education levels that are very low or within skilled trades.

**Cluster 4** is urban with established and affluent intellectuals and some western immigration. Its polling areas have a high share of people with higher education, Western immigrants and citizens, top executives and self-employed people, co-op housing and the gross top 20 percent income is high. It scores low on detached houses, social housing, welfare benefits, citizens from Eastern Europe and the Global South. The cluster contains most of Frederiksberg, Indre by, Christianshavn and Islands Brygge, central parts of Østerbro, some of Vesterbro, and a single polling area in Nørrebro that borders the lakes on one side and Østerbro on the other.

The polling area 'Indre By, Indre by' (the most central part of Copenhagen) is part of this cluster but labeled as an outlier, because, relatively to other areas in its cluster, it has more crime (arrests), is more male-dominated, has more citizens from Western countries and less Danish citizens, and also has more people on social benefits, lower voter turnout, and more home renters.

## Load data

```python
df = pd.read_csv(STANDARDISED)
id_cols = ['PollingAreaID','PollingArea']
polling_areas = df[id_cols]
df_standardised = df.drop(columns=id_cols)
df_num = pd.read_csv(NOT_STANDARDISED).drop(columns=id_cols)
```

## Models

```python
# Run PCA again for clustering
n_components = 5 # chosen number of principal components
pca_reduced = PCA(n_components=n_components)
X_pca = pca_reduced.fit_transform(df_standardised)
```

### Determine number of clusters

```python
inertias = []
silhouette_scores = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_pca)
    inertias.append(kmeans.inertia_)
    score = silhouette_score(X_pca, kmeans.labels_)
    silhouette_scores.append(score)

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(K, inertias, 'bo-')
ax1.set_xlabel('Number of clusters')
ax1.set_ylabel('Inertia')
ax1.set_title('Elbow Method')

ax2.plot(K, silhouette_scores, 'ro-')
ax2.set_xlabel('Number of clusters')
ax2.set_ylabel('Silhouette score')
ax2.set_title('Silhouette Analysis')
filename = 'elbow_method'
plt.savefig(f"{OUTPUT_PATH_PNG}{filename}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{OUTPUT_PATH_PGF}{filename}.pgf")
plt.savefig(f"{OUTPUT_PATH_PDF}{filename}.pdf")
```

```python
# Use the KneeLocator from the kneed library to find the elbow point
from kneed import KneeLocator
kl = KneeLocator(K, inertias, curve="convex", direction="decreasing")
elbow_point = kl.elbow
print(f"The optimal number of clusters is: {elbow_point}")
```

### k-Means

```python
k = elbow_point
kmeans = KMeans(
    n_clusters=k,
    init="k-means++",
    n_init=50,
    random_state=42
)

kmeans_labels = kmeans.fit_predict(X_pca)
centroids = kmeans.cluster_centers_

X_plot = X_pca[:, :2]
centroids_plot = centroids[:, :2]

# Store clusters
df_clusters = df_standardised.copy()
df_clusters['Cluster_kMeans'] = kmeans_labels

# Store PCA coordinates
for i in range(X_pca.shape[1]):
    df_clusters[f'PC{i+1}'] = X_pca[:, i]

# Store centroids separately
df_centroids = pd.DataFrame(
    centroids,
    columns=[f'PC{i+1}' for i in range(centroids.shape[1])]
)
df_centroids['Cluster_kMeans'] = range(1, k + 1)

# Map centroids back to original feature space
centroids_original_space = pca_reduced.inverse_transform(centroids)
df_centroids_original = pd.DataFrame(
    centroids_original_space,
    columns=df_standardised.columns
)
df_centroids_original['Cluster_kMeans'] = range(1, k + 1)
```

### Hierarchical clustering

```python
hierarchical = AgglomerativeClustering(n_clusters=elbow_point, linkage='ward')
hierarchical_labels = hierarchical.fit_predict(X_pca)
df_clusters['Cluster_Hierarchical'] = hierarchical_labels
```

### GMM

```python
gmm = GaussianMixture(n_components=k, covariance_type='full', random_state=42, n_init=10)
gmm_labels = gmm.fit_predict(X_pca)
gmm_probs = gmm.predict_proba(X_pca)
df_clusters['Cluster_GMM'] = gmm_labels
```

## Plot clusters

```python
# Use first 2 PCs for visualization
X_viz = X_pca[:, :2]
area_names = polling_areas['PollingArea'].values.tolist()

# Calculate variance explained by first 2 PCs
pc1_var = pca_reduced.explained_variance_ratio_[0]
pc2_var = pca_reduced.explained_variance_ratio_[1]

xlabel = f'PC1 ({pc1_var:.1%} variance)'
ylabel = f'PC2 ({pc2_var:.1%} variance)'
```

```python

def plot_data(X: np.ndarray, 
              labels: np.ndarray, 
              centroids: np.ndarray = None,
              show_distance: bool = False, 
              point_labels = None,
              xlabel: str = "Feature 1", 
              ylabel: str = "Feature 2", 
              title: str = "Clustering",
              filename: str = "plot",
              square: bool = False):
    """
    Plot the clustered data.

    Args:
        X (np.ndarray): The data points.
        labels (np.ndarray): The cluster labels.
        centroids (np.ndarray, optional): The centroids of the clusters.
            Defaults to `None`.
        show_distance (bool, optional): Whether to show the distance between
            the data points and the centroids. Defaults to `False`.
        square (bool, optional): Whether to make the plot square and centred 
            around 0 on both axes. Defaults to `False`.
    """

    plt.figure(figsize=(10, 8))

    # Show the distance between the data points and the centroids.
    if show_distance and centroids is not None:
        for centroid in centroids:
            for point in X:
                plt.plot(
                    [centroid[0], point[0]], [centroid[1], point[1]],
                    color=COLOURS[-1], alpha=0.3, linestyle="--",
                    linewidth=0.5
                )

    unique_labels = np.unique(labels)
    for label in unique_labels:
        if label == -1:
            # Outliers in black with 'x' marker
            mask = labels == label
            plt.scatter(
                X[mask, 0], X[mask, 1], 
                c='black', marker='x', s=100, 
                label='Outliers', alpha=0.8, linewidth=2
            )
        else:
            mask = labels == label
            color_idx = int(label) % len(COLOURS)
            plt.scatter(
                X[mask, 0],
                X[mask, 1], 
                c=COLOURS[color_idx], 
                s=80, 
                edgecolor=COLOURS[-1],
                alpha=0.6,
                label=f'Cluster {label}'
            )

    if point_labels is not None:
        for i, txt in enumerate(point_labels):
            plt.annotate(
                txt,
                (X[i, 0], X[i, 1]),
                fontsize=8,
                alpha=0.6,
                xytext=(3, 3),
                textcoords="offset points"
            )

    if centroids is not None:
        for i, centroid in enumerate(centroids):
            color_idx = i % (len(COLOURS) - 1)  # Match cluster color
            plt.scatter(centroid[0], centroid[1], 
                    s=180,              
                    c=COLOURS[color_idx],  # Use cluster color
                    marker='x',         
                    linewidth=2.5,       
                    alpha=0.8,          
                    zorder=5)
        
        # Add single legend entry for all centroids
        plt.scatter([], [], s=80, c='gray', marker='x', 
                linewidth=2.5, alpha=0.9, label='Centres')

    # Plot information.
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    # plt.axhline(0, color=COLOURS[-1], linewidth=0.8, alpha=0.5)
    # plt.axvline(0, color=COLOURS[-1], linewidth=0.8, alpha=0.5)
    plt.legend(loc='best', fontsize=10)
    plt.grid(False)

    # Make plot square and centred around 0 if requested
    if square:
        # Get current axis limits
        ax = plt.gca()
        x_min, x_max = ax.get_xlim()
        y_min, y_max = ax.get_ylim()
        
        # Find the maximum absolute value across both axes
        max_range = max(abs(x_min), abs(x_max), abs(y_min), abs(y_max))
        
        # Set symmetric limits centred on 0
        plt.xlim(-max_range, max_range)
        plt.ylim(-max_range, max_range)
        plt.gca().set_aspect('equal', adjustable='box')

    plt.savefig(f"{OUTPUT_PATH_PNG}{filename}.png", dpi=300, bbox_inches='tight')
    plt.savefig(f"{OUTPUT_PATH_PGF}{filename}.pgf")
    plt.savefig(f"{OUTPUT_PATH_PDF}{filename}.pdf")
    plt.show()
```

### Use eta-squared to find most important features

```python
# ============================================================================
# ETA-SQUARED: PROPORTION OF VARIANCE EXPLAINED BY CLUSTERS
# ============================================================================

def calculate_eta_squared(feature_values, cluster_labels):
    """
    Calculate eta-squared: proportion of variance explained by cluster membership.
    Range: 0 (no separation) to 1 (perfect separation)
    """
    # Total variance
    ss_total = np.sum((feature_values - feature_values.mean())**2)
    
    # Between-cluster variance
    cluster_means = [feature_values[cluster_labels == c].mean() 
                     for c in np.unique(cluster_labels)]
    cluster_sizes = [np.sum(cluster_labels == c) 
                     for c in np.unique(cluster_labels)]
    
    ss_between = sum(size * (mean - feature_values.mean())**2 
                     for size, mean in zip(cluster_sizes, cluster_means))
    
    # Eta-squared
    eta_squared = ss_between / ss_total if ss_total > 0 else 0
    
    return eta_squared

eta_squared_values = {}

for col in df_standardised.columns:
    feature_values = df_analysis[col].values
    cluster_labels = df_analysis['cluster'].values
    
    eta_squared = calculate_eta_squared(feature_values, cluster_labels)
    eta_squared_values[col] = eta_squared

# Sort by eta-squared
eta_df = pd.DataFrame.from_dict(eta_squared_values, orient='index', columns=['Eta-squared'])
eta_df = eta_df.sort_values('Eta-squared', ascending=False)

print("\n" + "="*80)
print("TOP FEATURES BY ETA-SQUARED (% variance explained by clusters)")
print("="*80)
print(f"{'Feature':<50} {'Eta²':>10} {'Interpretation'}")
print("-"*80)

for idx, row in eta_df.iterrows():
    eta_val = row['Eta-squared']
    
    # Interpretation
    if eta_val > 0.14:
        interpretation = "Strong"
    elif eta_val > 0.06:
        interpretation = "Moderate"
    elif eta_val > 0.01:
        interpretation = "Small"
    else:
        interpretation = "Negligible"
    
    print(f"{idx:<50} {eta_val:>10.3f} {interpretation:>15}")
```

### Explain outliers

```python
best_model = 'Cluster_kMeans'

# Start fresh with only numeric data
df_full = df_standardised.copy()
df_full['Cluster'] = df_clusters[best_model].values

# Store area names separately (don't mix with numeric data)
area_names = polling_areas['PollingArea'].values

# Get cluster profiles (means)
cluster_profiles = df_full.groupby('Cluster').mean()

# Get list of numeric columns (everything except 'Cluster')
numeric_cols = [col for col in df_full.columns if col != 'Cluster']

print(f"Analysing {len(outliers_kmeans)} outliers...")
print(f"Using {len(numeric_cols)} numeric features")

# Analyze each outlier
for idx in outliers_kmeans.index:
    area_name = area_names[idx]
    
    print(f"\n{'='*60}")
    print(f"OUTLIER: {area_name}")
    print(f"{'='*60}")
    
    # Get cluster assignment
    cluster_id = df_full.loc[idx, 'Cluster']
    print(f"Assigned to Cluster {cluster_id}")
    
    # Get cluster mean for this cluster
    cluster_mean = cluster_profiles.loc[cluster_id, numeric_cols]
    
    # Get outlier's values
    outlier_values = df_full.loc[idx, numeric_cols]
    
    # Calculate absolute differences
    differences = (outlier_values - cluster_mean).abs()
    
    # Sort and get top 10
    top_differences = differences.sort_values(ascending=False).head(10)
    
    print("\nTop 10 features where this area differs most from its cluster:")
    print(f"{'Feature':<50} {'Difference':<10}")
    print(f"{'-'*60}")
    for feature, diff in top_differences.items():
        outlier_val = outlier_values[feature]
        cluster_val = cluster_mean[feature]
        print(f"{feature:<50} {diff:>8.3f}")
        print(f"  → Outlier: {outlier_val:>7.3f} | Cluster mean: {cluster_val:>7.3f}")
```
