```python
import pandas as pd

path_1 = '../../../processed-data/elections/relative_2009.csv' 
path_2 = '../../../processed-data/elections/relative_2013.csv'
path_3 = '../../../processed-data/elections/relative_2017.csv'
path_4 = '../../../processed-data/elections/relative_2021.csv'

election_2009 = pd.read_csv(path_1)
election_2013 = pd.read_csv(path_2)
election_2017 = pd.read_csv(path_3)
election_2021 = pd.read_csv(path_4)

election_2009["year"] = 2009
election_2013["year"] = 2013
election_2017["year"] = 2017
election_2021["year"] = 2021

election_2009.head()
```

```python
elections = pd.concat(
    [election_2009, election_2013, election_2017, election_2021],
    ignore_index=True
)

elections.head()
```

```python
path_5 = '../../../processed-data/migration/relative_citizenship_over20years_2009.csv' 
path_6 = '../../../processed-data/migration/relative_citizenship_over20years_2013.csv'
path_7 = '../../../processed-data/migration/relative_citizenship_over20years_2017.csv'
path_8 = '../../../processed-data/migration/relative_citizenship_over20years_2021.csv'

migration_2009 = pd.read_csv(path_5)
migration_2013 = pd.read_csv(path_6)
migration_2017 = pd.read_csv(path_7)
migration_2021 = pd.read_csv(path_8)

migration_2009["year"] = 2009
migration_2013["year"] = 2013
migration_2017["year"] = 2017
migration_2021["year"] = 2021

migration_2009.head()
```

```python
migrations = pd.concat(
    [migration_2009, migration_2013, migration_2017, migration_2021],
    ignore_index=True
)

migrations.head()
```

```python
merged = elections.merge(
    migrations,
    left_on=["PollingAreaID", "year"],
    right_on=["ValgstedId", "year"],
    how="left"   # change if needed
)

merged = merged.drop(columns=["ValgstedId"])
merged.columns
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Make sure key columns are numeric
merged["Eligible voters"] = pd.to_numeric(merged["Eligible voters"], errors="coerce")
merged["Voter turnout"]  = pd.to_numeric(merged["Voter turnout"], errors="coerce")
merged["Valid votes"]    = pd.to_numeric(merged["Valid votes"], errors="coerce")
merged["Citizenship_Denmark"] = pd.to_numeric(merged["Citizenship_Denmark"], errors="coerce")

# Non-Danish share
merged["Non Danish Share"] = 100 - merged["Citizenship_Denmark"]

# Basic sanity check
print(merged[["Non Danish Share"]].describe())
print(merged[["Voter turnout"]].describe())
```

### Turnout distribution over time

```python
plt.figure()
merged.boxplot(column="Voter turnout", by="year")
plt.title("Voter turnout by election year")
plt.suptitle("")
plt.xlabel("Election year")
plt.ylabel("Voter turnout (%)")
plt.show()
```

This figure shows the distribution of voter turnout across polling areas for each election year. Median turnout increases noticeably between 2009 and 2013 and remains relatively stable in the following elections. While turnout becomes more evenly distributed across polling areas in the earlier period, later elections show greater variation, with several polling areas consistently recording lower turnout. This pattern suggests meaningful differences across polling areas and motivates further exploration of whether these differences are associated with variation in population composition.

```python
years = sorted(merged["year"].dropna().unique())

for y in years:
    sub = merged.loc[merged["year"] == y, "Voter turnout"].dropna()

    plt.figure()
    plt.hist(sub, bins=12)
    plt.title(f"Turnout distribution across polling areas ({y})")
    plt.xlabel("Voter turnout (%)")
    plt.ylabel("Number of polling areas")
    plt.show()
```

The turnout distributions show a clear rightward shift between 2009 and later elections, indicating higher overall participation across polling areas. At the same time, the presence of low-turnout polling areas in every election year suggests that differences across polling areas persist over time. This pattern points to stable underlying factors contributing to turnout variation.

```python
# Compute turnout quartile per year
merged["turnout_q25"] = (
    merged.groupby("year")["Voter turnout"]
          .transform(lambda x: x.quantile(0.25))
)

# Flag low-turnout polling areas
merged["low_turnout"] = merged["Voter turnout"] <= merged["turnout_q25"]
```

```python
low_turnout_counts = (
    merged[merged["low_turnout"]]
    .groupby("PollingAreaID")["year"]
    .count()
    .sort_values(ascending=False)
)

low_turnout_counts.head(10)
```

```python
low_turnout_summary = (
    merged[merged["low_turnout"]]
    .groupby("PollingAreaID")
    .agg(
        low_turnout_years=("year", "count"),
        avg_turnout=("Voter turnout", "mean"),
        avg_non_danish_share=("Non Danish Share", "mean")
    )
    .sort_values("low_turnout_years", ascending=False)
)

low_turnout_summary.head(10)
```

Identifying polling areas that consistently fall in the lowest turnout quartile reveals that low turnout is persistent in a subset of polling areas. These areas also vary in terms of population composition, with some exhibiting notably higher non-Danish population shares. This observation motivates further investigation into the relationship between population composition and voter turnout.

```python
# Count how many times each polling area is low-turnout
low_counts = (
    merged[merged["low_turnout"]]
    .groupby("PollingAreaID")["year"]
    .count()
)

# Identify consistently low-turnout areas
consistent_low_ids = low_counts[low_counts == 4].index

merged["consistent_low_turnout"] = merged["PollingAreaID"].isin(consistent_low_ids)

comparison = (
    merged
    .groupby("consistent_low_turnout")["Non Danish Share"]
    .mean()
    .rename(index={True: "Consistently low-turnout areas",
                   False: "Other polling areas"})
)

print(comparison)

```

```python
plt.figure()
comparison.plot(kind="bar")
plt.ylabel("Average non-Danish share (%)")
plt.title("Average non-Danish share by turnout persistence")
plt.xticks(rotation=0)
plt.show()
```

Polling areas that consistently exhibit low turnout across elections tend to have higher non-Danish population shares. On average, these areas have a non-Danish share of approximately 22%, compared to around 15% in other polling areas. Although this descriptive comparison does not imply causality, it highlights a systematic difference in population composition that may be relevant for understanding persistent turnout differences.

```python
turnout_wide = (
    merged.pivot_table(index="PollingAreaID", columns="year", values="Voter turnout", aggfunc="mean")
)

sub = turnout_wide[[2009, 2021]].dropna()

plt.figure(figsize=(6,6))
plt.scatter(sub[2009], sub[2021], alpha=0.7)

# 45-degree line
mn = min(sub[2009].min(), sub[2021].min())
mx = max(sub[2009].max(), sub[2021].max())
plt.plot([mn, mx], [mn, mx])

plt.xlabel("Turnout (%) in 2009")
plt.ylabel("Turnout (%) in 2021")
plt.title("Polling-area turnout stability (2009 vs 2021)")
plt.tight_layout()
plt.show()
```

The strong correlation between turnout in 2009 and 2021 suggests that spatial differences in voter participation are highly persistent. Despite changes in overall turnout levels, polling areas maintain their relative positions over time, pointing to enduring local characteristics rather than short-term demographic change as key drivers of turnout differences.

### Non-Danish share distribution over time

```python
plt.figure()
merged.boxplot(column="Non Danish Share", by="year")
plt.title("Non-Danish share by election year")
plt.suptitle("")
plt.xlabel("Election year")
plt.ylabel("Non-Danish share (%)")
plt.show()
```

The distribution of non-Danish population shares shifts upward over time, reflecting a gradual increase in diversity across polling areas. However, this change is uneven, with a subset of polling areas consistently exhibiting much higher non-Danish shares than others.

```python
for y in years:
    sub = merged.loc[merged["year"] == y, "Non Danish Share"].dropna()

    plt.figure()
    plt.hist(sub, bins=12)
    plt.title(f"Non-Danish share distribution across polling areas ({y})")
    plt.xlabel("Non-Danish share (%)")
    plt.ylabel("Number of polling areas")
    plt.show()
```

The histograms show that while non-Danish population shares increase over time, most polling areas remain clustered at relatively low to moderate levels. A small number of polling areas consistently exhibit much higher shares, highlighting uneven changes in population composition across space.

### Core association — Non-Danish share vs turnout

```python
for y in years:
    sub = merged[merged["year"] == y].dropna(subset=["Non Danish Share", "Voter turnout"])

    x = sub["Non Danish Share"].to_numpy()
    y_turn = sub["Voter turnout"].to_numpy()

    plt.figure()
    plt.scatter(x, y_turn)

    # Trend line
    if len(sub) >= 2:
        m, b = np.polyfit(x, y_turn, 1)
        xx = np.linspace(x.min(), x.max(), 100)
        plt.plot(xx, m * xx + b)

    plt.title(f"Non-Danish share vs voter turnout ({y})")
    plt.xlabel("Non-Danish share (%)")
    plt.ylabel("Voter turnout (%)")
    plt.show()
```

Across all election years, a negative association between non-Danish population share and voter turnout is observed at the polling-area level. Although the relationship is consistent in direction, turnout varies considerably within similar levels of population composition, highlighting the heterogeneity underlying this association.

```python
# Sort for correct differencing
tmp = merged.sort_values(["PollingAreaID", "year"]).copy()

# Ensure numeric
tmp["Non Danish Share"] = pd.to_numeric(tmp["Non Danish Share"], errors="coerce")
tmp["Voter turnout"] = pd.to_numeric(tmp["Voter turnout"], errors="coerce")
tmp["year"] = pd.to_numeric(tmp["year"], errors="coerce")

# Compute within-polling-area changes between elections
tmp["d_non_danish"] = tmp.groupby("PollingAreaID")["Non Danish Share"].diff()
tmp["d_turnout"] = tmp.groupby("PollingAreaID")["Voter turnout"].diff()

# Create interval label using shift (aligned)
tmp["year_prev"] = tmp.groupby("PollingAreaID")["year"].shift(1)
tmp["interval"] = tmp["year_prev"].astype("Int64").astype(str) + "-" + tmp["year"].astype("Int64").astype(str)

# Drop rows where diffs/interval aren't defined
dsub = tmp.dropna(subset=["d_non_danish", "d_turnout", "year_prev"])

plt.figure()

for interval in sorted(dsub["interval"].unique()):
    sub = dsub[dsub["interval"] == interval]

    # Small jitter
    x_jitter = sub["d_non_danish"] + np.random.normal(0, 0.1, size=len(sub))

    plt.scatter(
        x_jitter,
        sub["d_turnout"],
        label=interval,
        alpha=0.7
    )

plt.axhline(0)
plt.axvline(0)
plt.xlabel("Δ Non-Danish share (percentage points)")
plt.ylabel("Δ Voter turnout (percentage points)")
plt.title("Changes in non-Danish share vs changes in turnout, by election interval")
plt.legend(title="Election interval")
plt.show()
```

This figure examines the relationship between changes in non-Danish population share and changes in voter turnout between consecutive elections at the polling-area level. After accounting for limited variation in population composition in some election intervals (2013-2017), no clear systematic relationship emerges between changes in non-Danish share and changes in turnout. Across all election intervals, changes in turnout occur in both directions for similar changes in population composition, indicating substantial heterogeneity across polling areas. This suggests that short-term demographic changes within polling areas are not strongly associated with contemporaneous changes in voter turnout, and that the negative association observed in level-based analyses is more likely driven by persistent differences between polling areas rather than by short-term change.

```python
df = merged.copy()
df["Non Danish Share"] = pd.to_numeric(df["Non Danish Share"], errors="coerce")
df["Voter turnout"] = pd.to_numeric(df["Voter turnout"], errors="coerce")

# Create quintiles WITHIN each year so bins are comparable across years
df["nd_quintile"] = df.groupby("year")["Non Danish Share"].transform(
    lambda x: pd.qcut(x, 5, labels=False, duplicates="drop") + 1
)

binned = (
    df.dropna(subset=["nd_quintile", "Voter turnout"])
      .groupby(["year", "nd_quintile"])["Voter turnout"]
      .mean()
      .reset_index()
)

plt.figure(figsize=(7,5))
for y in sorted(binned["year"].unique()):
    sub = binned[binned["year"] == y]
    plt.plot(sub["nd_quintile"], sub["Voter turnout"], marker="o", label=str(int(y)))

plt.xlabel("Non-Danish share quintile (within year)")
plt.ylabel("Mean turnout (%)")
plt.title("Mean turnout by non-Danish share quintile (lines by year)")
plt.xticks([1,2,3,4,5])
plt.legend(title="Year")
plt.tight_layout()

plt.show()


```

### Citizenship

```python
citizenship_cols = [
    'Citizenship_Nordic',
    'Citizenship_Turkey',
    'Citizenship_Former Yugoslavia',
    'Citizenship_Old EU countries',
    'Citizenship_New EU countries',
    'Citizenship_Other Europe',
    'Citizenship_Africa',
    'Citizenship_North America',
    'Citizenship_South and Central America',
    'Citizenship_Asia and Oceania'
]

years = sorted(merged["year"].unique())

corr_df = pd.DataFrame(index=citizenship_cols, columns=years, dtype=float)

for col in citizenship_cols:
    for y in years:
        sub = merged[merged["year"] == y][[col, "Voter turnout"]].dropna()
        if len(sub) >= 5:
            corr_df.loc[col, y] = sub[col].corr(sub["Voter turnout"])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.heatmap(
    corr_df.astype(float),
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation between citizenship group shares and voter turnout\nby election year")
plt.xlabel("Election year")
plt.ylabel("Citizenship group")
plt.tight_layout()
plt.show()
```

For several groups; citizens from Africa, Asia and Oceania, Turkey, the former Yugoslavia, and New EU countries—the correlation with turnout is strongly negative and highly stable over time. In contrast, groups such as citizens from Old EU countries, Other Europe, and South and Central America exhibit weak or near-zero correlations. Nordic and North American citizens show a consistently positive association with turnout. Overall, the heatmap highlights both the stability of these patterns across elections and substantial heterogeneity across citizenship groups.

### Party

```python
party_cols = ['A','B','C','D','E','F','G','H','I','J','K','L','M','N','O','P','Q','R','T','U','V','Y','Z','Ø','Å','Æ','1','2','3']
party_cols = [p for p in party_cols if p in merged.columns]

years = sorted(merged["year"].unique())

corr_party = pd.DataFrame(index=party_cols, columns=years, dtype=float)

for p in party_cols:
    for y in years:
        sub = merged[merged["year"] == y][[p, "Valid votes", "Non Danish Share"]].dropna()
        if len(sub) >= 5:
            vote_share = sub[p] / sub["Valid votes"]
            corr_party.loc[p, y] = vote_share.corr(sub["Non Danish Share"])
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_party.astype(float),
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Correlation between party vote shares and non-Danish share\nby election year")
plt.xlabel("Election year")
plt.ylabel("Party")
plt.tight_layout()
plt.show()
```
